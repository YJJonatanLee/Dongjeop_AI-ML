# SigLIP Inference 예제

파인튜닝된 SigLIP 모델로 이미지 분류를 수행하는 예제입니다.

```python
import sys
from pathlib import Path

# 프로젝트 루트를 path에 추가
project_root = Path().resolve().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from infer.infer import SigLIPInference
```

## 모델 초기화

파인튜닝된 체크포인트를 사용하여 추론 객체를 생성합니다.

```python
# 체크포인트 경로 설정

# Linear Head 모델 (기본)
checkpoint_path = "../models/checkpoints/siglip-base-patch16-224/best_model_siglip-base-patch16-224_20251202_004235.pth"

# ML Decoder 모델 (권장!) - ML Decoder는 자동으로 감지됩니다
# checkpoint_path = "../models/checkpoints/siglip-base-patch16-224-mldecoder/best_model_*.pth"

# SigLIP2를 사용하려면:
# checkpoint_path = "../models/checkpoints/siglip2-base-patch16-224/best_model_siglip2-base-patch16-224_20251202_024457.pth"

# 모델 ID 설정 (필요시)
# model_id = "google/siglip-base-patch16-224"
# model_id = "google/siglip2-base-patch16-224"

# 추론 객체 생성
# ML Decoder 체크포인트를 사용하면 자동으로 ML Decoder 모델이 로드됩니다
inferencer = SigLIPInference(
    checkpoint_path=checkpoint_path,
    threshold=0.5,
    verbose=False
)

print(f"사용 중인 레이블: {inferencer.label_names}")
```

## 추론

추론 객체에 이미지 url과 threshold를 입력하여 result를 구합니다.

```python
# URL 이미지
image_url = "https://media-cdn.tripadvisor.com/media/photo-s/11/11/6b/cf/caption.jpg"

# 추론 실행 (threshold 조정 가능)
result = inferencer.infer(image_url, threshold=0.5)

print("추론 결과:", result)
```
